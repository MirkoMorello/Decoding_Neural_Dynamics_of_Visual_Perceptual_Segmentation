```python
import matplotlib.pyplot as plt
import numpy as np

# --- Data Preparation ---
tasks = ['SALICON', 'MIT1003', 'MIT1003 Scanpath']

# LL scores for each of the four models across the three tasks
deepgaze_scores = [0.76, 2.06, 2.62]
deepgaze_spade_scores = [0.81, 2.14, 2.65] # Added model. NOTE: 2.65 is a placeholder!
dinogaze_scores = [0.82, 2.20, 2.70]
dinogaze_spade_scores = [0.82, 2.19, 2.73]

# --- Plotting Logic ---
x = np.arange(len(tasks))  # the label locations
width = 0.20  # the width of the bars, made narrower to fit four

fig, ax = plt.subplots(figsize=(14, 8)) # Made figure slightly larger

# Define a color scheme to group the backbones
# Blues for CNN-based models, Oranges/Reds for ViT-based models
color_cnn_base = 'steelblue'
color_cnn_spade = 'dodgerblue'
color_vit_base = 'darkorange'
color_vit_spade = '#3a9f1b'

# Create bars for each model with new offsets for a group of four
rects1 = ax.bar(x - 1.5*width, deepgaze_scores, width, label='DeepGazeIII (CNN)', color=color_cnn_base)
rects2 = ax.bar(x - 0.5*width, deepgaze_spade_scores, width, label='DG-SPADE v3 (CNN)', color=color_cnn_spade)
rects3 = ax.bar(x + 0.5*width, dinogaze_scores, width, label='DinoGaze (ViT)', color=color_vit_base)
rects4 = ax.bar(x + 1.5*width, dinogaze_spade_scores, width, label='DinoGaze-SPADE (ViT)', color=color_vit_spade)

# --- Aesthetics and Labels ---
ax.set_ylabel('Log-Likelihood (LL)', fontsize=12)
ax.set_title('Full Model Comparison: Log-Likelihood Across Tasks', fontsize=16, weight='bold')
ax.set_xticks(x)
ax.set_xticklabels(tasks, fontsize=11)
ax.legend(fontsize=11)
ax.set_ylim(0, max(dinogaze_spade_scores) * 1.1)

# Function to add data labels on top of each bar
def add_bar_labels(rects):
    """Attach a text label above each bar in *rects*, displaying its height."""
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.2f}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),  # 3 points vertical offset
                    textcoords="offset points",
                    ha='center', va='bottom', weight='bold')

add_bar_labels(rects1)
add_bar_labels(rects2)
add_bar_labels(rects3)
add_bar_labels(rects4)

# Add a subtle grid and remove top/right spines for a cleaner look
ax.grid(axis='y', linestyle='--', alpha=0.6)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

fig.tight_layout()
plt.show()
```
